```python
df1 = pd.DataFrame({'ID':[1,2,2,3],'Score':[88,96,77,79]})
df1
```

```python
df2 = pd.DataFrame({'ID':[1,2,3],'Name':['A','B','C']})
df2
```

```python
#Vertical / row level / top on top
pd.concat([df1,df2],axis=0)
```

```python
#Horizontal / row level / top on top
pd.concat([df1,df2],axis=1)
```

```python
#merge = Join
pd.merge(df1,df2, how='inner',on='ID')
```

```python
pd.merge(df1, df2, how='inner',left_on='ID', right_on='ID')
```
